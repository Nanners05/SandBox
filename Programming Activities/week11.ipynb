```python
import pandas as pd

df = pd.read_csv("/workspaces/SandBox/lecture_notes/titanic.csv")
```

```python
df = df.drop(columns=["Cabin"])
df["Age"] = df["Age"].fillna(df["Age"].median())
df = df.dropna(subset=["Embarked"])
```

```python
df = df.rename(columns={"Pclass": "Passenger_Class"})
df = df.rename(columns={"SibSp": "Siblings/Spouses"})
df = df.rename(columns={"Parch": "Parent/Children"})
df.rename(columns={"PassengerID": "Passenger_ID"})
```

```python
"""
Programming Activity 1

Using the Titanic dataset and the Pandas library, fill in any missing data cells in the 'ages' column with the median age.
"""
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
"""
Programming Activity 2

Using the Titanic dataset and the Pandas library, create a new column called 'FamilySize' and fill it with the size of the passenger's family.
"""
df["FamilySize"] = df["Siblings/Spouses"] + df["Parent/Children"]
```

```python
"""
Programming Activity 3

Using the Titanic dataset and the Pandas library, find the average survival rate by class and sex.
"""
survival_rate = df.groupby(["Passenger_Class", "Sex"])["Survived"].mean()
print("Avg survival rate by class and sex", survival_rate)
```

```python
"""
Programming Activity 4

Using the Titanic dataset and the Pandas library, find the average fare per class.
"""
fare_per_class = df.groupby(["Passenger_Class"])["Fare"].mean()
print("Avg fare per class:", fare_per_class)
```
